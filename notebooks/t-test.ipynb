```python
import numpy as np
from scipy.stats import t

def independent_t_test(group1, group2, alpha=0.05):
    # Step 1: Compute sample sizes
    n1 = len(group1)
    n2 = len(group2)

    # Step 2: Compute means
    mean1 = np.mean(group1)
    mean2 = np.mean(group2)

    # Step 3: Compute variances
    var1 = np.var(group1, ddof=1)  # Sample variance
    var2 = np.var(group2, ddof=1)

    # Step 4: Compute t-statistic (Welch's t-test, no equal variance assumption)
    t_stat = (mean1 - mean2) / np.sqrt(var1 / n1 + var2 / n2)

    # Step 5: Compute degrees of freedom using Welch-Satterthwaite equation
    df = ((var1 / n1 + var2 / n2) ** 2) / (
        ((var1 / n1) ** 2) / (n1 - 1) + ((var2 / n2) ** 2) / (n2 - 1)
    )

    # Step 6: Compute p-value (two-tailed test)
    p_value = 2 * (1 - t.cdf(abs(t_stat), df))

    # Step 7: Determine significance
    if p_value < alpha:
        result = "Reject the null hypothesis: The means are significantly different."
    else:
        result = "Fail to reject the null hypothesis: The means are not significantly different."

    # Output results
    print(f"Group 1 Mean: {mean1:.2f}, Group 2 Mean: {mean2:.2f}")
    print(f"Group 1 Variance: {var1:.2f}, Group 2 Variance: {var2:.2f}")
    print(f"T-Statistic: {t_stat:.4f}, Degrees of Freedom: {df:.2f}")
    print(f"P-Value: {p_value:.4f}")
    print(result)



def independent_t_test_from_summary(mean1, std1, n1, mean2, std2, n2, alpha=0.05):
    # Step 1: Compute variances from standard deviations
    var1 = std1 ** 2
    var2 = std2 ** 2

    # Step 2: Compute t-statistic (Welch's t-test, no equal variance assumption)
    t_stat = (mean1 - mean2) / np.sqrt(var1 / n1 + var2 / n2)

    # Step 3: Compute degrees of freedom using Welch-Satterthwaite equation
    df = ((var1 / n1 + var2 / n2) ** 2) / (
        ((var1 / n1) ** 2) / (n1 - 1) + ((var2 / n2) ** 2) / (n2 - 1)
    )

    # Step 4: Compute p-value (two-tailed test)
    p_value = 2 * (1 - t.cdf(abs(t_stat), df))

    # Step 5: Determine significance
    if p_value < alpha:
        result = "Reject the null hypothesis: The means are significantly different."
    else:
        result = "Fail to reject the null hypothesis: The means are not significantly different."

    # Output results
    print(f"Group 1 Mean: {mean1:.2f}, Group 2 Mean: {mean2:.2f}")
    print(f"Group 1 Std: {std1:.2f}, Group 2 Std: {std2:.2f}")
    print(f"T-Statistic: {t_stat:.4f}, Degrees of Freedom: {df:.2f}")
    print(f"P-Value: {p_value:.4f}")
    print(result)





```

```python
independent_t_test_from_summary(29.6, 0.76, 4, 29.95, 1.05, 4)
```
